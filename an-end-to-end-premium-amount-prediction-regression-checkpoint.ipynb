### Why Convert to Sine and Cosine?

* Date-related features like months or days are cyclical. For example:

    * After December (month 12), the next is January (month 1), forming a cycle.
    * Similarly, days repeat every 31 days (or the number of days in a month).
* If you encode these as integers (`1, 2, ..., 12`), machine learning models may incorrectly assume a linear relationship, e.g., December (`12`) is farther from January (`1`) than November (`11`).

* Sine and cosine transform these features into a circular representation, ensuring that December and January are close in the feature space.

 * Example:

     **For months:**

    * January (`Month = 1`) → (sin, cos) = (0.5, 0.866)

    * December (`Month = 12`) → (sin, cos) = (0.5, 0.866)


```python
df_temp['Year_Sin'] = np.sin(2 * np.pi * df_temp['StartYear'])
df_temp['Year_Cos'] = np.cos(2 * np.pi * df_temp['StartYear'])
df_temp['Month_Sin'] = np.sin(2 * np.pi * df_temp['StartMonth'] / 12) 
df_temp['Month_Cos'] = np.cos(2 * np.pi * df_temp['StartMonth'] / 12)
df_temp['Day_Sin'] = np.sin(2 * np.pi * df_temp['StartDay'] / 31)  
df_temp['Day_Cos'] = np.cos(2 * np.pi * df_temp['StartDay'] / 31)
```

This line creates a custom "grouping" feature that captures the time progression in weeks:
* (`Year - 2020`) scales everything relative to the year 2020.
* `Month * 4` aligns months approximately with weeks (since 12 months × ~4 weeks = 48 weeks per year).
* `Day // 7` groups days into weekly intervals.


**Purpose:**
* This feature may help models learn trends or behaviors that vary with time.

```python
df_temp['dateGroup'] = (df_temp['StartYear'] - 2020) * 48 + df_temp['StartMonth'] * 4 + df_temp['StartDay'] // 7
```

Now we've checked the individual distributions, we will take a look at making a scatter plot with each column against the target column: `Premium Amount`. Using the Age column to group the data and create a hue of colours as the age increases


```python
# Define the function to draw subplots
def plot_scatterplots_optimized(data, target, hue_column, columns, sample_size=None):
    n_cols = 3  # Number of subplots per row
    n_rows = (len(columns) + n_cols - 1) // n_cols  # Calculate rows based on the number of columns

    # Downsample the data (optional)
    if sample_size and len(data) > sample_size:
        data = data.sample(sample_size, random_state=42)

    # Create multiple subplots
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 5))
    axes = axes.flatten()  # Flatten the axes for easier access

    for i, column in enumerate(columns):
        sns.scatterplot(ax=axes[i], data=data, x=column, y=target, hue=hue_column, alpha=0.7)
        axes[i].set_title(f'{column} vs {target}', fontsize=14)
        axes[i].set_xlabel(column, fontsize=10)
        axes[i].set_ylabel(target, fontsize=10)
        axes[i].legend(title=hue_column, loc='best')
        axes[i].grid(True)

    # Hide unused subplots
    for j in range(len(columns), len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout(pad = 4.5)
    plt.show()
```

### Evaluation function Creation

Our Evaluation will be based off of RMSLE (Root Mean Square Log Error), sklearn does not have this metric, so we will convert our predictions and training labels (y) to log values using the `np.log1p` function and use the available RMSE (Root Mean Square Error) evaluation function

```python
from sklearn.metrics import mean_squared_error
```

```python
def rmsle(y_true,y_pred):
    log_true = np.log1p(y_true)
    log_pred = np.log1p(y_pred)
    return np.sqrt(mean_squared_error(log_true,log_pred))
```

```python
def preprocess_data(df):
    """
    Perform transformation on df and returns transformed df.
    """
    # Fill the numeric rows with median
    for label, content in df.items():
        if pd.api.types.is_numeric_dtype(content):
            if pd.isnull(content).sum():
                # Fill missing numeric values with median
                df[label] = content.fillna(content.median())
            
            # Fill non-numerical missing data
        if not pd.api.types.is_numeric_dtype(content):
            if pd.isnull(content).sum():
                # Fill missing numeric values with median
                df[label] = content.fillna('Missing')
    
    return df
```

## Hyperparameter tuning using GridSearchCV

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid for GridSearchCV
param_grid = {
    'max_depth': [None, 3, 5, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'n_estimators': [3002],
    'subsample': [0.7, 0.8, 1.0, None],
    'colsample_bytree': [0.7, 0.8, 1.0, None],
    'gamma': [0.0, 0.1, 0.3],
    'random_state' : [42]
}

# Initialize XGBoost regression model with additional parameters
model = xg.XGBRegressor(
    objective='reg:squarederror',  # Regression with squared error loss function
    seed=42,                      # For reproducibility
    learning_rate=0.01,            # Base learning rate for fine-tuning
    eval_metric='rmse'             # Monitor RMSE during training
)
```

```python
# Wrap the XGBoost model in a GridSearchCV
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring='neg_mean_squared_error',
    cv=3, # 3-fold cross-validation
    n_jobs = 1
)
```

```python
# Perform Grid Search

# Custom fit function with early stopping
grid_search.fit(
    x_train, y_train,
    eval_set=[(x_val, y_val)],
    # eval_metric='rmse',
    # early_stopping_rounds=70,  # Early stopping applied for each fold
    verbose=3
)
```

## Prediction on test set given

we will create a function to perform all operations we performed on our train set on our test set

```python
def preprocess(df):
    """
    This Function processes data and puts it in the format of the training set 
    for predictions by our machine learning 
    """
    # Sort DataFrame in date order
    df.sort_values(by = ["Policy Start Date"], inplace = True, ascending = True)
    df["Policy Start Date"].head(20)

    # Add datetime Parameters for policy start date
    df["StartYear"] = df["Policy Start Date"].dt.year
    df["StartMonth"] = df["Policy Start Date"].dt.month
    df["StartDay"] = df["Policy Start Date"].dt.day
    df["StartDayOfWeek"] = df["Policy Start Date"].dt.dayofweek
    df["StartDayofYear"] = df["Policy Start Date"].dt.dayofyear

    df["StartMicrosecond"] = df["Policy Start Date"].dt.microsecond
    
    # Convert Date and time variables to sine and cosine for cyclic representation
    df['Year_Sin'] = np.sin(2 * np.pi * df['StartYear'])
    df['Year_Cos'] = np.cos(2 * np.pi * df['StartYear'])
    df['Month_Sin'] = np.sin(2 * np.pi * df['StartMonth'] / 12) 
    df['Month_Cos'] = np.cos(2 * np.pi * df['StartMonth'] / 12)
    df['Day_Sin'] = np.sin(2 * np.pi * df['StartDay'] / 31)  
    df['Day_Cos'] = np.cos(2 * np.pi * df['StartDay'] / 31)
    df['dateGroup'] = (df['StartYear'] - 2020) * 48 + df['StartMonth'] * 4 + df['StartDay'] // 7

    # Now we've enriched our DataFrame with date time features, we can remove Policy Start Date, date only and and every other columns with only one unique value
    df = df.drop(columns =["Policy Start Date", "id"], axis = 1)

    # Demographic-based features
    df['IsMarried'] = df['Marital Status'].apply(lambda x: 1 if x == 'Married' else 0)
    df['HasDependents'] = df['Number of Dependents'].apply(lambda x: 1 if x > 0 else 0)
    df['IsSmoker'] = df['Smoking Status'].apply(lambda x: 1 if x == 'Yes' else 0)
    df['does_exercises'] = df['Exercise Frequency'].apply(lambda x: 1 if x in ['Daily', 'Weekly'] else 0)
    
    # Financial-based features
    df['ClaimsRisk'] = df['Previous Claims'].apply(lambda x: 1 if x > 0 else 0)
    
    # Interaction Features
    df['Age_SmokingRisk'] = (df['Age'].fillna(df['Age'].median())) * df['IsSmoker']
    df["CreditScore_HealthScore"] = (df["Credit Score"].fillna(df['Credit Score'].median())) * df['Health Score']
    
                                           
    df['Income_to_dependant_ratio'] = (df['Annual Income'].fillna(df['Annual Income'].median())) / (df['Number of Dependents'].fillna(0) + 1)  # Adding 1 to avoid division by zero 
    # Also we are filling missing values in the dependant column by asssuming Number of Dependents missing is 0
    
    df['Health_Risk_score'] = df['IsSmoker'] + df['does_exercises'] + (100 - df['Health Score'])/100

    # Number of dependants columns is filled with 0 assuming that missing values in that column mean no dependants
    df['Number of Dependants'] = df['Number of Dependents'].fillna(0)

    # Filling in missing values in other columns
    # Fill the numeric rows with median
    for label, content in df.items():
        if pd.api.types.is_numeric_dtype(content):
            if pd.isnull(content).sum():
                # Fill missing numeric values with median
                df[label] = content.fillna(content.median())
            
            # Fill non-numerical missing data with 'Missing'
        if not pd.api.types.is_numeric_dtype(content):
            if pd.isnull(content).sum():
                # Fill missing numeric values with median
                df[label] = content.fillna('Missing')

    # Convert categorical data to numerical data
    cat_colm = ['Gender', 'Customer Feedback', 'Smoking Status', 'Marital Status', 'Education Level']
    for i in cat_colm:
        df[i] = le.fit_transform(df[i])

    # This will turn all of the remaining object value one hot encoding using pd.get_dummies
    # Loop through each column in the dataframe
    for label, content in df.items():
        if pd.api.types.is_object_dtype(content):
           # Use pd.get_dummies() to convert object column into dummy variables
           dummies = pd.get_dummies(content, prefix=label, drop_first=False)
           # Drop the original object column and concatenate the dummies
           df = pd.concat([df, dummies], axis=1).drop(label, axis=1)

    return df
```
